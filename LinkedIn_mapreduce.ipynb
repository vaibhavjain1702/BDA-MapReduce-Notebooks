```python
# Reduce phase


def reduce_views(shuffled_data):
    reduced_data = {}
    for key, values in shuffled_data.items():
        # Sum up the values for each content_id (reduce step)
        reduced_data[key] = sum(values)
    return reduced_data
```

# Counting Word Frequencies Across Web Pages

```python
# Input: Web Pages (Text Data)

# Sample web pages data (list of strings where each string represents the content of a web page)
web_pages = [
    "MapReduce is a programming model for processing large data sets.",
    "MapReduce distributes the processing over a cluster of computers.",
    "MapReduce uses two main functions: map and reduce."
]
```

```python
# Map Phase

def map_words(web_pages):
    mapped_data = []
    for page in web_pages:
        words = page.split()  # Split page content into words
        for word in words:
            # Emit (word) as key, and 1 as value
            mapped_data.append((word.lower(), 1))  # Convert to lowercase for uniformity
    return mapped_data
```

```python
# Shuffle and Sort Phase

from collections import defaultdict

def shuffle_and_sort(mapped_data):
    shuffled_data = defaultdict(list)
    for key, value in mapped_data:
        shuffled_data[key].append(value)
    return shuffled_data
```

```python
# Reduce Phase

def reduce_word_count(shuffled_data):
    reduced_data = {}
    for key, values in shuffled_data.items():
        # Sum the values (reduce step)
        reduced_data[key] = sum(values)
    return reduced_data
```

```python
# Step 1: Map Phase
mapped_data = map_words(web_pages)

# Step 2: Shuffle and Sort Phase
shuffled_data = shuffle_and_sort(mapped_data)

# Step 3: Reduce Phase
reduced_data = reduce_word_count(shuffled_data)

# Output the final result
for word, count in reduced_data.items():
    print(f"Word: {word}, Count: {count}")
```
